```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
df = pd.read_excel(nama_file)

kolom_butir = [col for col in df.columns if "[" in col]
butir = df[kolom_butir]

n = len(df)
jumlah_butir = butir.shape[1]

print(f"Jumlah responden (n): {n}")
print(f"Jumlah butir pernyataan: {jumlah_butir}")
print(f"Data kosong (NaN): {int(butir.isna().sum().sum())}")
```

```python
skala = {
"Sangat Tidak Setuju (STS)": 1,
"Tidak Setuju (TS)": 2,
"Netral (N)": 3,
"Setuju (S)": 4,
"Sangat Setuju (SS)": 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")
# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")
print(f"Sel yang gagal dikonversi: {int(num.isna().sum().sum())}")
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
  # Menggunakan regex untuk mengambil nama dimensi sebelum tanda '['
  match = re.match(r"^(.*?)\s*\[", kolom)
  nama = match.group(1).strip() if match else "Unknown"
  dimensi.setdefault(nama, []).append(kolom)

print("\n--- Ringkasan Dimensi ---")
for nama, daftar in dimensi.items():
  print(f"{nama}: {len(daftar)} butir")
```

```python
dimensi = {}
for kolom in butir.columns:
  # Menggunakan regex untuk mengambil nama dimensi sebelum tanda '['
  match = re.match(r"^(.*?)\s*\[", kolom)
  nama = match.group(1).strip() if match else "Unknown"
  dimensi.setdefault(nama, []).append(kolom)

print("\n--- Ringkasan Dimensi ---")
for nama, daftar in dimensi.items():
  print(f"{nama}: {len(daftar)} butir")
```

```python
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"\nNilai r tabel (alpha=0.05, df={n-2}) = {r_tabel:.3f}\n")

hasil_validitas = []
for nama, daftar in dimensi.items():
  x = num[daftar]
  total = x.sum(axis=1)
  for i, kolom in enumerate(daftar, 1):
    r, p = stats.pearsonr(x[kolom], total)
    r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
    hasil_validitas.append({
        "Dimensi": nama,
        "Butir": f"B{i}",
        "r hitung": round(r, 3),
        "Corrected r": round(r_koreksi, 3),
        "p-value": round(p, 4),
        "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
    })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

```python
def cronbach_alpha(x):
  k = x.shape[1]
  var_sum = x.var(ddof=1).sum()
  total_var = x.sum(axis=1).var(ddof=1)

  # Mencegah error pembagian dengan nol jika varians total adalah 0
  if total_var == 0:
    return 0
  return k / (k - 1) * (1 - var_sum / total_var)

def kategori(a):
  if a >= 0.90:
    return "Sangat baik"
  elif a >= 0.80:
    return "Baik"
  elif a >= 0.70:
    return "Reliabel"
  else:
    return "Tidak reliabel"

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
  x = num[daftar]
  a = cronbach_alpha(x)
  hasil_reliabilitas.append({
      "Dimensi": nama,
      "Jumlah butir": len(daftar),
      "Cronbach's alpha": round(a, 3),
      "Kategori": kategori(a),
  })

  for i, kolom in enumerate(daftar, 1):
    # Butir hanya dihitung jika dimensi memiliki lebih dari 2 butir
    if len(daftar) > 2:
      alpha_dihapus.append({
          "Dimensi": nama,
          "Butir": f"B{i}",
          "Alpha jika dihapus": round(
              cronbach_alpha(x.drop(columns=kolom)), 3
          ),
      })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left"
)
with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
  gabung.to_excel(w, sheet_name="Validitas", index=False)
  tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```
